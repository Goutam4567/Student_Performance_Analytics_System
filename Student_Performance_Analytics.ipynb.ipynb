```python
print("Number of students:", len(df))
print("\nColumns:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()
```

```python
print("Basic Statistics:")
display(df.describe())
```

```python
subjects = ["Python", "NumPy", "Pandas", "AI"]
print("Subjects:", subjects)
```

```python
df["Total"] = df[subjects].sum(axis=1)
print("Total marks calculated:")
display(df[["Student_ID", "Name", "Total"]])
```

```python
df["Average"] = df[subjects].mean(axis=1)
print("Average marks calculated:")
display(df[["Student_ID", "Name", "Average"]])
```

```python
def assign_grade(average):
 if average >= 90:
  return "A+"
 elif average >= 80:
  return "A"
 elif average >= 70:
  return "B"
 elif average >= 60:
  return "C"
 elif average >= 50:
  return "D"
 else:
  return "F"
df["Grade"] = df["Average"].apply(assign_grade)

print("Grades assigned:")
display(df[["Student_ID", "Name", "Average", "Grade"]])
```

```python
def pass_fail(row):
    marks = row[subjects].to_numpy(dtype=float)

    if np.all(marks >= 40):
        return "Pass"
    else:
        return "Fail"

df["Result"] = df.apply(pass_fail, axis=1)

print("Pass/Fail Status:")
display(df[["Student_ID", "Name", "Result"]])
```

```python

from IPython.display import display

print("Complete Student Analysis:")

display(
    df[
        [
            "Student_ID",
            "Name",
            "Department",
            "Python",
            "NumPy",
            "Pandas",
            "AI",
            "Attendance",
            "Total",
            "Average",
            "Grade",
            "Result"
        ]
    ]
)
```

```python
import numpy as np

print("===== CLASS SUMMARY =====")

print("Number of students:", len(df))
print("Overall class average:", round(np.mean(df["Average"]), 2))
print("Highest average:", round(np.max(df["Average"]), 2))
print("Lowest average:", round(np.min(df["Average"]), 2))
print("Passed students:", (df["Result"] == "Pass").sum())
print("Failed students:", (df["Result"] == "Fail").sum())
```

```python
subjects = ["Python", "NumPy", "Pandas", "AI"]

subject_avg = df[subjects].mean()

print("===== SUBJECT-WISE AVERAGE =====")
display(subject_avg.round(2))

print("Best subject:", subject_avg.idxmax())
print("Lowest average subject:", subject_avg.idxmin())
```

```python
subjects = ["Python", "NumPy", "Pandas", "AI"]
marks_array = df[subjects].to_numpy()

print("===== NUMPY ANALYSIS =====")
print("Overall marks average:", round(np.mean(marks_array), 2))
print("Highest individual subject mark:", np.max(marks_array))
print("Lowest individual subject mark:", np.min(marks_array))
```

```python
top_students = df.nlargest(3, "Average")[["Student_ID", "Name", "Average", "Grade"]]

print("===== TOP 3 PERFORMERS =====")
display(top_students)
```

```python
grade_distribution = df["Grade"].value_counts().sort_index()

print("===== GRADE DISTRIBUTION =====")
display(grade_distribution)
```

```python
result_distribution = df["Result"].value_counts()

print("===== PASS / FAIL DISTRIBUTION =====")
display(result_distribution)
```

```python
highest_student = df.loc[df["Average"].idxmax()]

print("===== HIGHEST PERFORMER =====")
print("Student ID:", highest_student["Student_ID"])
print("Name:", highest_student["Name"])
print("Average:", round(highest_student["Average"], 2))
print("Grade:", highest_student["Grade"])
```

```python
lowest_student = df.loc[df["Average"].idxmin()]

print("===== LOWEST PERFORMER =====")
print("Student ID:", lowest_student["Student_ID"])
print("Name:", lowest_student["Name"])
print("Average:", round(lowest_student["Average"], 2))
print("Grade:", lowest_student["Grade"])
```
